```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Prepare the data
data = {
    'Scale': [0.5, 0.75, 1.0, 1.25, 1.5],
    'loss': [1.882, 2.136, 1.604, 2.175, 2.185],
    'c_loss': [1.456, 0.989, 0.495, 1.024, 1.385],
    'f_loss': [0.412, 1.130, 1.104, 1.135, 0.785],
    'm_loss': [0.013, 0.016, 0.005, 0.016, 0.015],
    'PIR': [0.059, 0.631, 0.998, 0.788, 0.235],
    'IR': [0.017, 0.496, 0.966, 0.684, 0.169],
    'RRE': [126.069, 27.574, 1.246, 15.965, 90.960],
    'RTE': [1.140, 0.443, 0.024, 0.248, 1.182],
    'RMSE': [0.444, 0.239, 0.017, 0.262, 1.240],
    'RR': [0.000, 0.666, 1.000, 0.832, 0.062]
}

df = pd.DataFrame(data)

# 2. Setup the plot grid
metrics = [col for col in df.columns if col != 'Scale']
n_metrics = len(metrics)
fig, axes = plt.subplots(nrows=4, ncols=3, figsize=(15, 18))
axes = axes.flatten()

# 3. Iterate and plot each metric
for i, metric in enumerate(metrics):
    ax = axes[i]
    ax.plot(df['Scale'], df[metric], marker='o', linestyle='-', linewidth=2, color='#1f77b4')
    ax.set_title(f'Scale vs {metric}', fontsize=14, fontweight='bold')
    ax.set_xlabel('Scale Factor', fontsize=10)
    ax.set_ylabel(metric, fontsize=10)
    ax.grid(True, linestyle='--', alpha=0.6)
    
    # Highlight the baseline (Scale 1.0)
    ax.axvline(x=1.0, color='red', linestyle=':', alpha=0.5, label='Baseline')

# Hide any empty subplots in the grid
for j in range(i + 1, len(axes)):
    axes[j].axis('off')

plt.tight_layout()
plt.savefig('scale_analysis.png', dpi=300)
plt.show()
```

```python
# downsample and view point cloud

import os
import numpy as np

np.random.seed(42)

src = "UHM_generated_data/0.ply"
dst_dir = "UHM_downsampled"
dst = os.path.join(dst_dir, "0.ply")

os.makedirs(dst_dir, exist_ok=True)

N = count
downsample_ratio = 0.85
keep = int(N*(1-downsample_ratio))
idx = np.random.choice(N, keep, replace = False)

pcd_downsampled = pcd.select_by_index(idx)

o3d.io.write_point_cloud(dst, pcd_downsampled)
o3d.visualization.draw_geometries([pcd_downsampled])
print(keep)
```

```python
# pick_open3d.py
import os
import json
import open3d as o3d

src = "data/UHM_downsampled/train/0.ply"      # file you want to pick from
out_idx_path = "picked_indices.json"

pcd = o3d.io.read_point_cloud(src)

vis = o3d.visualization.VisualizerWithEditing()
vis.create_window(window_name="Pick points: left-click to pick, close window to finish")
vis.add_geometry(pcd)

opt = vis.get_render_option()
opt.point_size = 5.0              # make points easier to click
vis.run()                         # left-click to pick points; close window when finished
picked = vis.get_picked_points()
vis.destroy_window()

print("Picked indices:", picked)

with open(out_idx_path, "w") as f:
    json.dump(picked, f)
print(f"Saved {len(picked)} indices to {out_idx_path}")
```

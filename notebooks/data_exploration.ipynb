### Load sample 

```python
DATA_ROOT = DATA_FOLDER / "3DMatch" / "data"    
```

```python
train_example = train_metadata[0]
```

```python
src_points = torch.load(DATA_ROOT / train_example["pcd0"])
```

```python
ref_points = torch.load(DATA_ROOT / train_example["pcd1"])
```

```python
ref_points
```

```python
ref_pc = o3d.geometry.PointCloud()
ref_pc.points = o3d.utility.Vector3dVector(ref_points)
ref_pc.colors = o3d.utility.Vector3dVector(
    np.tile(np.array([[1.0, 0.0, 0.0]]), (ref_points.shape[0], 1))
)
src_pc = o3d.geometry.PointCloud()
src_pc.points = o3d.utility.Vector3dVector(src_points)
src_pc.colors = o3d.utility.Vector3dVector(
    np.tile(np.array([[0.0, 1.0, 0.0]]), (src_points.shape[0], 1))
)
```

```python
o3d.visualization.draw_plotly([ref_pc, src_pc])
```
